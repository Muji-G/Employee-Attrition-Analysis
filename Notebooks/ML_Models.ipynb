# ML Models

```python
from sklearn.metrics import precision_score, recall_score

from Src.Data_Processing import (
    load_data,
    clean_data,
    feature_engineering,
    get_categorical_columns,
    encode_features
)

from Src.ML_Models import *

df = load_data("C:/Users/Admin/OneDrive/Documents/GitHub/Employee-Attrition-Analysis/Data/Raw/IBM HR Analysis.csv")
df = clean_data(df)
df = feature_engineering(df)
```

```python
# Identifying categorical columns
categorical_cols = get_categorical_columns(df)

# Ensuring target variable is not encoded
categorical_cols = [col for col in categorical_cols if col != "Attrition"]

# One-hot encoding
df_encoded = encode_features(df, categorical_cols)


# Train / test split
X_train, X_test, y_train, y_test, feature_names = prepare_data(
    df_encoded,
    target_col="Attrition"
)

# Checking class distribution
print("Train distribution (%):")
print(y_train.value_counts(normalize=True) * 100)

print("\nTest distribution (%):")
print(y_test.value_counts(normalize=True) * 100)
```

## Training, Evaluating models

```python
# Train
lr_model = train_logistic_regression(X_train, y_train)
dt_model = train_decision_tree(X_train, y_train)
rf_model = train_random_forest(X_train, y_train)


# Evaluate
evaluate(lr_model, X_train, X_test, y_train, y_test)
evaluate(dt_model, X_train, X_test, y_train, y_test)
evaluate(rf_model, X_train, X_test, y_train, y_test)


# Feature importance
show_feature_importance(lr_model, feature_names)
show_feature_importance(dt_model, feature_names)
show_feature_importance(rf_model, feature_names)
```

# Model comparison table

```python
models = [lr_model, dt_model, rf_model]
model_names = ["Logistic Regression", "Decision Tree", "Random Forest"]

results = []

for model, name in zip(models, model_names):
    y_pred = model.predict(X_test)
    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred)
    })

results_df = pd.DataFrame(results)
print("\n===== Model Comparison =====")
print(results_df)
```
