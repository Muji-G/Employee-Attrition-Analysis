```python
plt.figure(figsize=(15, 12))

numeric_df = df.select_dtypes(include=["int64", "float64"])

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    fmt=".2f",
    cmap="RdYlGn",
    annot_kws={"size": 12},
    linewidths=0.5,
    cbar_kws={"shrink": 0.8}
)

plt.title("Correlation Heatmap", fontsize=16)
plt.show()

plt.savefig("../reports/figures/Correlation Heatmap.png", dpi=300, bbox_inches="tight")
```

```python
numeric_df = df.select_dtypes(include=["int64", "float64"])

top_corr_features = (
    numeric_df.corr()
    .nlargest(20, "Attrition")["Attrition"]
    .index
)

plt.figure(figsize=(15, 15))
sns.heatmap(
    numeric_df[top_corr_features].corr(),
    annot=True,
    fmt=".2f",
    cmap="RdYlGn",
    annot_kws={"size": 10},
    linewidths=0.5
)
plt.title("Top 20 Features Correlated with Attrition", fontsize=16)
plt.show()
plt.savefig("../reports/figures/Top 20 Correlation.png", dpi=300, bbox_inches="tight")
```

```python
# Computing correlation of numeric features with Attrition
corr_with_attrition = (
    df.drop("Attrition", axis=1)
      .select_dtypes(include=["int64", "float64"])
      .corrwith(df["Attrition"])
      .sort_values()
)

# Converting to DataFrame
corr_df = corr_with_attrition.reset_index()
corr_df.columns = ["Feature", "Correlation"]

# Horizontal bar chart
plt.figure(figsize=(10, 8))
sns.barplot(
    data=corr_df,
    x="Correlation",
    y="Feature",
    hue="Feature",
    palette="RdYlGn",
    legend=False
)

plt.title("Correlation of Features with Attrition", fontsize=14)
plt.xlabel("Correlation")
plt.ylabel("Feature")
plt.show()
plt.savefig("../reports/figures/Correlation Horizontal Bar Chart.png", dpi=300, bbox_inches="tight")
```

Monthly income is highly correlated with Job level.

Job level is highly correlated with total working hours.

Monthly income is highly correlated with total working hours.

Age is also positively correlated with the Total working hours.

Marital status and stock option level are negatively correlated
